```python
class Number:
    def even(self):
        i = int(input("Enter a number:\n"))
        for x in range(0, i):
            if x % 2 == 0:
                print(x)

# Create instance and call method
num = Number()
num.even()
```

```python
import pandas as pd

# Step 1: Define the dataset
data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "Age": [25, 30, 35, 28, 40],
    "City": ["Kigali", "Musanze", "Huye", "Rubavu", "Nyagatare"]
}

# Step 2: Create a DataFrame
df = pd.DataFrame(data)

# Step 3: Display the DataFrame
print("Original DataFrame:")
print(df)

# Step 4: Manipulations

# Example 1: Add a new column (e.g., Age next year)
df["Age_next_year"] = df["Age"] + 1

# Example 2: Filter individuals older than 30
older_than_30 = df[df["Age"] > 30]

# Example 3: Select only Name and City columns
name_city = df[["Name", "City"]]

# Display results
print("\nDataFrame with Age_next_year:")
print(df)

print("\nIndividuals older than 30:")
print(older_than_30)

print("\nName and City columns:")
print(name_city)
```

```python
# Define the Car class
class Car:
    def __init__(self, brand, model):
        self.brand = brand  # Attribute for brand
        self.model = model  # Attribute for model

    def display_info(self):
        """Return a string with car details"""
        return f"Car Brand: {self.brand}, Model: {self.model}"

# Create an instance of Car
my_car = Car("Toyota", "Corolla")

# Print the car's details
print(my_car.display_info())
```
